```python

conn = sqlite3.connect("customer.db")

cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS customers (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT,
    age INTEGER
)
""")

# Sample data insert
cursor.execute("INSERT INTO customers (name, age) VALUES ('Parth', 21)")
cursor.execute("INSERT INTO customers (name, age) VALUES ('Rahul', 23)")

conn.commit()

# Fetch records
query = "SELECT * FROM customers"

df = pd.read_sql(query, conn)

print(df)

# Connection close
conn.close()
```
