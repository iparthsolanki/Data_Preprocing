Part D: Outlier Handling

```python
from scipy.stats import zscore

z_scores = np.abs(
    zscore(
        merged_df[
            [
                "annual_income",
                "loan_amount",
                "credit_score"
            ]
        ]
    )
)

outliers = (
    z_scores > 3
).any(axis=1)

merged_df_z = merged_df[
    ~outliers
]
```

### Q1 = merged_df[

```python
Q1 = merged_df[
    "annual_income"
].quantile(0.25)

Q3 = merged_df[
    "annual_income"
].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR

merged_df_iqr = merged_df[
    (
        merged_df["annual_income"]
        >= lower
    )
    &
    (
        merged_df["annual_income"]
        <= upper
    )
]
```

### low = merged_df[

```python
low = merged_df[
    "annual_income"
].quantile(0.01)

high = merged_df[
    "annual_income"
].quantile(0.99)

merged_df["annual_income"] = (
    merged_df[
        "annual_income"
    ].clip(low,high)
)
```

### from scipy.stats.mstats import winsorize

```python
from scipy.stats.mstats import winsorize

merged_df[
    "annual_income"
] = winsorize(
    merged_df["annual_income"],
    limits=[0.01,0.01]
)
```
